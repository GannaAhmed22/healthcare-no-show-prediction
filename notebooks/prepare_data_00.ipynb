```python
import pandas as pd

DATA_PATH = '../data/raw'


calendar = pd.read_csv(f'{DATA_PATH}/DS1_calendar_dim.csv')
facility = pd.read_csv(f'{DATA_PATH}/DS1_facility_dim.csv')
patient = pd.read_csv(f'{DATA_PATH}/DS1_patient_dim.csv')
payer = pd.read_csv(f'{DATA_PATH}/DS1_payer_dim (1).csv')
provider = pd.read_csv(f'{DATA_PATH}/DS1_provider_dim.csv')
appointments = pd.read_csv(f'{DATA_PATH}/DS2_appointments_clean (1).csv')
```

```python
drop_from_model = [
    'appointment_id',
    'appointment_created_datetime',
    'appointment_datetime',
    'appointment_date',
    'appointment_month',
    'check_in_datetime',
    'seen_datetime',
    'cancel_datetime',
    'attendance_outcome',
    'arrival_late_minutes',
    'wait_to_room_minutes',
    'total_visit_minutes'
]
appointments = appointments.drop(columns=drop_from_model)
appointments.columns.to_list()
```

```python
drop_features = [
    'month_name',
    'weekday_name'
]

calendar = calendar.drop(columns=drop_features)
calendar.columns.to_list()
```

```python
drop_features = [
    'region',
    'county_fips',
    'bed_count',
    'opening_year',
    'active_flag'
]

facility = facility.drop(columns=drop_features)
facility.columns.to_list()
```

```python
drop_features = [
    'facility_id',
    'active_flag'
]

provider = provider.drop(columns=drop_features)
provider.columns.to_list()
```

```python
drop_features = [
    'birth_date',
    'gender_identity_group',
    'portal_active_flag',
    'active_status',
    'death_date'
]

patient = patient.drop(columns=drop_features)
patient.columns.to_list()
```

```python
drop_features = [
    'plan_name',
    'product_type',
    'prior_auth_intensity',
    'avg_payment_days_band',
    'network_rules_group',
    'market_share_band',
    'pharmacy_benefit_flag',
    'active_flag'
]

payer = payer.drop(columns=drop_features)
payer.columns.to_list()
```

```python
import os

path = '../data/ready_to_use'


payer.to_csv(f'{path}/payers_cleaned.csv', index=False)
patient.to_csv(f'{path}/patients_cleaned.csv', index=False)
appointments.to_csv(f'{path}/appointments_cleaned.csv', index=False)
provider.to_csv(f'{path}/providers_cleaned.csv', index=False)
facility.to_csv(f'{path}/facilities_cleaned.csv', index=False)
calendar.to_csv(f'{path}/calendar_cleaned.csv', index=False)
```
