```python
import pandas as pd

path = "../data/ready_to_use"

facilities = pd.read_csv(f'{path}/facilities_cleaned.csv')
payers = pd.read_csv(f'{path}/payers_cleaned.csv')
calendar = pd.read_csv(f'{path}/calendar_cleaned.csv')
```

```python
print(facilities.shape)
facilities.info()

print(facilities.isnull().sum())
```

```python
print(payers.shape)
payers.info()

print(payers.isnull().sum())
```

```python
print(calendar.shape)
calendar.info()

print(calendar.isnull().sum())
calendar_dropped_col=['calendar_date','week_of_year']
calendar=calendar.drop(columns=calendar_dropped_col)
```

```python
facilities.to_csv(
    '../data/clean/facilities_cleaned.csv',
    index=False
)
payers.to_csv(
    '../data/clean/payers_cleaned.csv',
    index=False
)
calendar.to_csv(
    '../data/clean/calendar_cleaned.csv',
    index=False
)
```
