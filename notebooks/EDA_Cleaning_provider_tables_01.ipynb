```python
import pandas as pd
path = "../data/ready_to_use"

appointments = pd.read_csv(f'{path}/appointments_cleaned.csv')
providers = pd.read_csv(f'{path}/providers_cleaned.csv')
```

```python
providers.head()
```

```python
print(providers.shape)
providers.info()

print(providers.isnull().sum())
# print("Duplicates:", providers.duplicated().sum())
```

```python
providers['panel_size'].describe()
```

```python
missing_panel = providers[providers['panel_size'].isna()]

missing_panel['provider_id'].isin(appointments['provider_id']).sum()
missing_panel_droped = missing_panel['provider_id'].isin(appointments['provider_id']) == False
missing_panel_droped.sum()

```

```python
providers = providers.drop(columns=["language_2"])
providers = providers.drop(
    missing_panel_droped.index
)
```

```python
providers['panel_size'] = providers.groupby(
    ['provider_gender', 'provider_type']
)['panel_size'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
providers.info()
```

```python
providers.to_csv(
    '../data/clean/providers_cleaned.csv',
    index=False
)
```
