```python
conn = sqlite3.connect("healthcare.db")

# Enable foreign keys
conn.execute("PRAGMA foreign_keys = ON")

cursor = conn.cursor()

print("Healthcare database created!")
```

```python
query = """

SELECT
    a.appointment_id,
    a.appointment_date,

    p.preferred_language,
    p.sex_at_birth,

    pr.provider_type,
    pr.specialty_group,

    f.facility_name,

    py.payer_group,

    g.deprivation_quintile,

    c.quarter,
    c.weekend_flag,
    c.holiday_flag,

    a.lead_days,
    a.copay_usd,
    a.distance_to_clinic_miles,
    a.no_show_flag

FROM appointments a

LEFT JOIN patient p
    ON a.patient_id = p.patient_id

LEFT JOIN provider pr
    ON a.provider_id = pr.provider_id

LEFT JOIN facility f
    ON a.facility_id = f.facility_id

LEFT JOIN payer py
    ON a.payer_id = py.payer_id

LEFT JOIN geography g
    ON a.zip_area_id = g.zip_area_id

LEFT JOIN calendar c
    ON a.appointment_date = c.calendar_date

LIMIT 20;

"""

result = pd.read_sql_query(query, conn)

display(result)
```

1. Are appointment reminders more effective for certain patient groups based on age, preferred language, and portal activity?
This query compares no-show rates for patients who received reminders and those who did not, grouped by age, language, and portal activity.

The results will show attendance rate and no-show rate for each patient or appointment group. This helps identify which factors are associated with missed appointments.

```python
query1 = """
SELECT
    CASE
        WHEN a.patient_age < 18 THEN 'Under 18'
        WHEN a.patient_age BETWEEN 18 AND 30 THEN '18-30'
        WHEN a.patient_age BETWEEN 31 AND 45 THEN '31-45'
        WHEN a.patient_age BETWEEN 46 AND 60 THEN '46-60'
        ELSE 'Above 60'
    END AS age_group,

    p.preferred_language,
    a.reminder_sent_flag,

    COUNT(*) AS total_appointments,

    SUM(a.no_show_flag) AS total_no_shows,

    ROUND(
        100.0 * SUM(a.no_show_flag) / COUNT(*), 2
    ) AS no_show_rate,

    ROUND(
        100.0 * (COUNT(*) - SUM(a.no_show_flag)) / COUNT(*), 2
    ) AS attendance_rate

FROM appointments a

JOIN patient p
    ON a.patient_id = p.patient_id

GROUP BY
    age_group,
    p.preferred_language,
    a.reminder_sent_flag

ORDER BY age_group;
"""

df1 = pd.read_sql_query(query1, conn)

display(df1)
```

Business Question 2: Insurance and Socioeconomic Factors
How do insurance type, copayment amount, and geographical socioeconomic factors relate to appointment attendance?

```python
query2 = """
SELECT
    py.payer_group,

    CASE
        WHEN a.copay_usd = 0 THEN 'No Copayment'
        WHEN a.copay_usd <= 20 THEN 'Low Copayment'
        WHEN a.copay_usd <= 50 THEN 'Medium Copayment'
        ELSE 'High Copayment'
    END AS copay_category,

    g.deprivation_quintile,

    COUNT(*) AS total_appointments,
    SUM(a.no_show_flag) AS total_no_shows,

    ROUND(AVG(a.copay_usd), 2) AS average_copayment,

    ROUND(
        100.0 * SUM(a.no_show_flag) / COUNT(*), 2
    ) AS no_show_rate,

    ROUND(
        100.0 * (COUNT(*) - SUM(a.no_show_flag)) / COUNT(*), 2
    ) AS attendance_rate

FROM appointments a

LEFT JOIN payer py
    ON a.payer_id = py.payer_id

LEFT JOIN geography g
    ON a.zip_area_id = g.zip_area_id

GROUP BY
    py.payer_group,
    copay_category,
    g.deprivation_quintile

ORDER BY
    g.deprivation_quintile,
    py.payer_group;
"""

df2 = pd.read_sql_query(query2, conn)

display(df2)
```

Business Question 3: Previous No-Show History
Are patients with a history of missed appointments more likely to miss future appointments?

```python
query3 = """
SELECT
    CASE
        WHEN prior_no_show_count_12m = 0
            THEN 'No Previous No-Shows'
        WHEN prior_no_show_count_12m BETWEEN 1 AND 2
            THEN '1-2 Previous No-Shows'
        WHEN prior_no_show_count_12m BETWEEN 3 AND 5
            THEN '3-5 Previous No-Shows'
        ELSE 'More Than 5 Previous No-Shows'
    END AS previous_no_show_group,

    COUNT(*) AS total_appointments,
    SUM(no_show_flag) AS total_no_shows,

    ROUND(
        AVG(prior_no_show_count_12m), 2
    ) AS average_previous_no_shows,

    ROUND(
        100.0 * SUM(no_show_flag) / COUNT(*), 2
    ) AS no_show_rate,

    ROUND(
        100.0 * (COUNT(*) - SUM(no_show_flag)) / COUNT(*), 2
    ) AS attendance_rate

FROM appointments

GROUP BY previous_no_show_group

ORDER BY no_show_rate DESC;
"""

df3 = pd.read_sql_query(query3, conn)

display(df3)
```

Business Question 4: Previous No-Shows and Booking Lead Time
Are patients with a history of missed appointments more likely to miss appointments when booking lead times are longer?

```python
query4 = """
SELECT
    CASE
        WHEN prior_no_show_count_12m = 0
            THEN 'No Previous No-Shows'
        ELSE 'Previous No-Shows'
    END AS patient_history,

    CASE
        WHEN lead_days <= 7 THEN '0-7 Days'
        WHEN lead_days <= 30 THEN '8-30 Days'
        WHEN lead_days <= 60 THEN '31-60 Days'
        ELSE 'More Than 60 Days'
    END AS booking_lead_time,

    COUNT(*) AS total_appointments,
    SUM(no_show_flag) AS total_no_shows,

    ROUND(
        100.0 * SUM(no_show_flag) / COUNT(*), 2
    ) AS no_show_rate,

    ROUND(
        100.0 * (COUNT(*) - SUM(no_show_flag)) / COUNT(*), 2
    ) AS attendance_rate

FROM appointments

GROUP BY
    patient_history,
    booking_lead_time

ORDER BY
    patient_history,
    booking_lead_time;
"""

df4 = pd.read_sql_query(query4, conn)

display(df4)
```
